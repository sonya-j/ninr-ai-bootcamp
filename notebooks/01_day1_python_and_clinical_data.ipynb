## 5. Download the real clinical dataset

The next cell uses the `ucimlrepo` package to retrieve the UCI Diabetes 130-US Hospitals dataset. The original dataset contains 101,766 hospital encounters and 47 features.

For Day 1, we will create a smaller **teaching subset** with only the variables we need.

> This is real, publicly available research data — not simulated patient data.

```python
from ucimlrepo import fetch_ucirepo
from pathlib import Path

# Fetch the UCI dataset
diabetes = fetch_ucirepo(id=296)

# Combine the feature and target data
full = pd.concat([diabetes.data.features, diabetes.data.targets], axis=1)

print("Rows:", len(full))
print("Columns:", len(full.columns))
```

```python
# Variables we will use in today's beginner activities
keep_cols = [
    "encounter_id", "race", "gender", "age",
    "time_in_hospital", "num_lab_procedures", "num_procedures",
    "num_medications", "number_outpatient", "number_emergency",
    "number_inpatient", "A1Cresult", "diabetesMed", "readmitted"
]

# Keep a reproducible sample so the Day 1 file stays small
teaching = full[keep_cols].sample(n=5000, random_state=42).reset_index(drop=True)

Path("../data").mkdir(exist_ok=True)
teaching.to_csv("../data/diabetes_day1.csv", index=False)

print("Teaching rows:", len(teaching))
teaching.head()
```

## 12. Make a simple plot

A picture can help us understand the distribution of a variable quickly.

```python
import matplotlib.pyplot as plt

teaching["time_in_hospital"].hist()
plt.xlabel("Days in hospital")
plt.ylabel("Number of encounters")
plt.title("Length of Hospital Stay")
plt.show()
```
