## 9. Focus on a manageable set of variables
Real datasets can have many columns. Researchers often begin with a subset relevant to their question.

```python
focus = data[[
    "mean radius",
    "mean texture",
    "mean perimeter",
    "mean area",
    "diagnosis"
]]

focus.head()
```

## 10. Describe the data

```python
focus.describe()
```

### Explore 🟡
Select one column and calculate its mean. Change the column name and rerun.

```python
focus["mean radius"].mean()
```

## 11. Count categories

## 13. Filter rows
Suppose we want to look only at malignant observations.

```python
malignant = focus[focus["diagnosis"] == "malignant"]
malignant.head()
```

## 14. Compare groups
Now we can ask an exploratory question: how does mean radius differ by diagnosis?

```python
focus.groupby("diagnosis")["mean radius"].mean()
```

### Explore 🟡
Replace `mean radius` with `mean texture` or `mean area`. What changes?

```python
focus.groupby("diagnosis")["mean texture"].mean()
```

## 15. Make a first visualization

```python
focus["mean radius"].hist()
plt.xlabel("Mean radius")
plt.ylabel("Number of observations")
plt.title("Distribution of mean radius")
plt.show()
```

### Your turn 🟢
Change the graph to display `mean texture` instead.

```python
focus["mean texture"].hist()
plt.xlabel("Mean texture")
plt.ylabel("Number of observations")
plt.title("Distribution of mean texture")
plt.show()
```

## 16. Compare distributions visually

```python
focus.boxplot(column="mean radius", by="diagnosis")
plt.title("Mean radius by diagnosis")
plt.suptitle("")
plt.xlabel("Diagnosis")
plt.ylabel("Mean radius")
plt.show()
```

### Think like a scientist 🔵
What can this plot tell us? What can it **not** tell us? Does seeing a group difference establish causation or clinical usefulness?
